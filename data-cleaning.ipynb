# Alko wine price prediction

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Tyynekaisa/alko-hinnasto/refs/heads/main/data/alko.csv"

alko_data = pd.read_csv(url, sep=";", skiprows=3)

print(alko_data.columns)
display(alko_data.head())

alko_data = alko_data[
    alko_data["Tyyppi"].isin(["valkoviinit", "punaviinit", "roseeviinit"])
]

print(alko_data["Vuosikerta"].unique())

# Filtering bottle size

alko_wines_filtered = alko_data[alko_data["Pullokoko"] == "0.75 l"]

display(alko_wines_filtered.describe())

display(alko_wines_filtered.head(10))
```
